**Task 3 – Create SaaS Customer Class (OOP)**

```python
class SaaSCustomer:

    def __init__(self, customer_id, name, signup_date, plan, monthly_fee, usage_hours):
        self.customer_id = customer_id
        self.name = name
        self.signup_date = signup_date
        self.plan = plan
        self.monthly_fee = monthly_fee
        self.usage_hours = usage_hours

    def validate_data(self):

        # 1. Replace missing name
        if self.name is None or self.name.strip() == "":
            self.name = "Unknown Customer"

        # 2. Standardize plan name
        self.plan = self.plan.strip().capitalize()

        # 3. Convert monthly_fee to float
        try:
            self.monthly_fee = float(self.monthly_fee)
        except (ValueError, TypeError):
            self.monthly_fee = 0.0

        # 4. Convert usage_hours to integer
        try:
            self.usage_hours = int(self.usage_hours)
        except (ValueError, TypeError):
            self.usage_hours = 0

    def display_customer(self):
        print("Customer Name:", self.name)


# Create a customer
customer = SaaSCustomer(
    101,
    "Rahul Sharma",
    "2023-01-10",
    "pro",
    "49.99",
    "120"
)

# Validate the data
customer.validate_data()

# Display customer name
customer.display_customer()
```

**Task 4 – Data Cleaning Pipeline**

```python
from datetime import datetime


# 1. Handle Missing Values
def handle_missing_values(data):

    # Handle missing names
    for i in range(len(data["name"])):
        if data["name"][i] is None or data["name"][i] == "":
            data["name"][i] = "Unknown Customer"

    # Handle missing usage hours
    for i in range(len(data["usage_hours"])):
        if data["usage_hours"][i] is None or data["usage_hours"][i] == "NULL":
            data["usage_hours"][i] = 0

    # Handle invalid monthly fee
    for i in range(len(data["monthly_fee"])):
        try:
            data["monthly_fee"][i] = float(data["monthly_fee"][i])
        except (ValueError, TypeError):
            data["monthly_fee"][i] = None

    return data


# 2. Remove Duplicate Records
def remove_duplicates(data):

    seen_ids = set()
    duplicate_indexes = []

    for i in range(len(data["customer_id"])):

        customer_id = data["customer_id"][i]

        if customer_id in seen_ids:
            duplicate_indexes.append(i)
        else:
            seen_ids.add(customer_id)

    # Remove duplicate records
    for i in reversed(duplicate_indexes):
        for key in data:
            data[key].pop(i)

    return data


# 3. Standardize Date Formats
def standardize_dates(data):

    for i in range(len(data["signup_date"])):

        date_value = data["signup_date"][i]

        formats = [
            "%Y-%m-%d",
            "%Y/%m/%d",
            "%d-%m-%Y",
            "%Y%m%d"
        ]

        converted_date = None

        for date_format in formats:
            try:
                converted_date = datetime.strptime(
                    date_value, date_format
                ).strftime("%Y-%m-%d")
                break
            except ValueError:
                continue

        data["signup_date"][i] = converted_date

    return data


# Apply the functions
saas_data = handle_missing_values(saas_data)
saas_data = remove_duplicates(saas_data)
saas_data = standardize_dates(saas_data)


# Display cleaned data
for i in range(len(saas_data["customer_id"])):
    print(
        f"ID: {saas_data['customer_id'][i]} | "
        f"Name: {saas_data['name'][i]} | "
        f"Signup Date: {saas_data['signup_date'][i]} | "
        f"Plan: {saas_data['plan'][i]} | "
        f"Monthly Fee: {saas_data['monthly_fee'][i]} | "
        f"Usage Hours: {saas_data['usage_hours'][i]}"
    )
```

**Task 5 – Summary Statistics and Insights**

```python
# Total customers
total_customers = len(saas_data["customer_id"])

# Monthly fee
valid_fees = []

for fee in saas_data["monthly_fee"]:
    try:
        valid_fees.append(float(fee))
    except (ValueError, TypeError):
        pass

total_monthly_fee = sum(valid_fees)
average_monthly_fee = total_monthly_fee / len(valid_fees)

# Usage hours
valid_usage = []

for hours in saas_data["usage_hours"]:
    try:
        valid_usage.append(int(hours))
    except (ValueError, TypeError):
        valid_usage.append(0)

total_usage_hours = sum(valid_usage)

# Plans
plans = saas_data["plan"]

# Print results
print("Total Customers:", total_customers)
print("Average Monthly Fee:", round(average_monthly_fee, 2))
print("Total Usage Hours:", total_usage_hours)

print("\nPlan Distribution")

for plan in set(plans):
    print(f"{plan} = {plans.count(plan)}")
```

**Task 6 – Identify Low-Usage Customers**

```python
def find_low_usage_customers(data):

    print("Low Usage Customers")

    for i in range(len(data["customer_id"])):

        usage = int(data["usage_hours"][i])

        if usage < 50:
            print(
                f"ID: {data['customer_id'][i]} | "
                f"Name: {data['name'][i]} | "
                f"Plan: {data['plan'][i]} | "
                f"Usage Hours: {usage}"
            )


find_low_usage_customers(saas_data)
```

**Task 8 – Save Cleaned Data to File**

```python
import csv

# -------------------------------
# 1. Save cleaned data to CSV
# -------------------------------

try:
    with open("saas_customers_cleaned.csv", "w", newline="") as file:

        writer = csv.writer(file)

        # Header
        writer.writerow([
            "customer_id",
            "name",
            "signup_date",
            "plan",
            "monthly_fee",
            "usage_hours"
        ])

        # Customer records
        for i in range(len(saas_data["customer_id"])):

            writer.writerow([
                saas_data["customer_id"][i],
                saas_data["name"][i],
                saas_data["signup_date"][i],
                saas_data["plan"][i],
                saas_data["monthly_fee"][i],
                saas_data["usage_hours"][i]
            ])

except Exception as e:
    print("Error while writing CSV file:", e)

else:
    print("CSV file created successfully.")

finally:
    print("CSV file operation completed.")


# -------------------------------
# 2. Create plan_summary.txt
# -------------------------------

try:
    plans = saas_data["plan"]

    with open("plan_summary.txt", "w") as file:

        for plan in sorted(set(plans)):
            count = plans.count(plan)
            file.write(f"{plan} : {count}\n")

except Exception as e:
    print("Error while writing plan summary:", e)

else:
    print("Plan summary file created successfully.")

finally:
    print("Plan summary operation completed.")
```
